```python
import pandas as pd

df = pd.read_csv(
    "C:/EcoWatt-AI/data/processed/final_energy_anomaly_results.csv"
)

print(df.shape)
```

```python
if "final_predicted_anomaly" in df.columns:
    df["anomaly"] = df["final_predicted_anomaly"]

if "final_anomaly_score" in df.columns:
    df["anomaly_score"] = df["final_anomaly_score"]
```

```python
normal_df = df[df["anomaly"] == 0]

anomaly_df = df[df["anomaly"] == 1]

print(len(normal_df))
print(len(anomaly_df))
```

```python
normal_sample = normal_df.sample(
    n=24000,
    random_state=42
)

anomaly_sample = anomaly_df.sample(
    n=6000,
    random_state=42
)
```

```python
dashboard_df = pd.concat(
    [normal_sample, anomaly_sample],
    ignore_index=True
)
```

```python
dashboard_df = dashboard_df.sample(
    frac=1,
    random_state=42
).reset_index(drop=True)
```

```python
dashboard_df.to_csv(
    "C:/EcoWatt-AI/data/processed/dashboard_dataset.csv",
    index=False
)
```
